```python
import pandas as pd 
import numpy as np
from pathlib import Path

#This approach is useful when your project contains multiple data files, you define the base directory once and resuse it 
DATA_DIR = Path(r"D:\SQL Project\retail-sales-rfm-analysis\data")
xlsx_path = DATA_DIR /"raw"/"online_retail_II.xlsx"
```

```python
#loading both the sheets together 
sheets = pd.read_excel(xlsx_path,sheet_name=None,engine ="openpyxl")
print(sheets.keys())

for name , d in sheets.items():
    print (name,d.shape)                 # sheet name and the shape of  the data frame 
```

```python
#Combining the both sheets into a single data frame 
df_0910 = sheets["Year 2009-2010"].assign(source_sheet="2009-2010")
df_1011 = sheets["Year 2010-2011"].assign(source_sheet="2010-2011")

df = pd.concat([df_0910, df_1011], ignore_index=True)

df.to_csv(DATA_DIR /"raw"/"online_retail_II_Combined.csv",index = False)
df.shape
```

```python
# understanding the data 
print (df.info())
print(df.isna().sum())     # null count in each column
print((df.isna().mean()*100).round(2))     # null % per column
print(df.describe())                       # numeric summery
print(df["InvoiceDate"].min(),df["InvoiceDate"].max())    #date range 

print(df.groupby("source_sheet")["InvoiceDate"].agg(["min","max","count"]))  #date range and count per sheet
# also yahan se date overlapping pakad aa gyi 

# how many invoices starts with C (Cancelled invoices)
print(df["Invoice"].astype(str).str.startswith("C").sum())  # count of cancelled invoices 
```
